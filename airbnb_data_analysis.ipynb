```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style for better aesthetics
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 8)

# 1.1 Load the Airbnb dataset into a Pandas DataFrame.
file_path = '1730285881-Airbnb_Open_Data (1) (1).xlsx'
df = pd.read_excel(file_path)

# 1.2 Display the first 5 rows of the dataset.
print("--- First 5 Rows of the Dataset ---")
display(df.head())

# 1.3 Get a concise summary of the dataframe.
print("\n--- DataFrame Info ---")
df.info()

# 1.4 Check for missing values in each column.
print("\n--- Missing Values per Column ---")
print(df.isnull().sum())


# 2.1 Handle missing values in 'reviews per month' by filling them with 0.
df['reviews per month'].fillna(0, inplace=True)
print("Missing values in 'reviews per month' after filling:", df['reviews per month'].isnull().sum())

# 2.2 Drop the 'license' column.
# The 'license' column is completely empty, so it's safe to drop.
df.drop('license', axis=1, inplace=True)
print("\n'license' column dropped.")

# 2.3 Drop rows with missing values in 'last review' and 'review rate number'.
# These columns are crucial for review-based analysis.
df.dropna(subset=['last review', 'review rate number'], inplace=True)
print("Rows with missing 'last review' or 'review rate number' dropped.")

# 2.4 Convert the 'last review' column to datetime objects.
df['last review'] = pd.to_datetime(df['last review'])
print("\n'last review' column converted to datetime.")

# 2.5 Clean 'price' and 'service fee' columns.
# Remove '$' and ',' characters and convert to numeric (float).
for col in ['price', 'service fee']:
    # Check if the column is of object type before trying string operations
    if df[col].dtype == 'object':
        df[col] = df[col].str.replace('$', '').str.replace(',', '').astype(float)

# Also, drop any rows where 'price' or 'service fee' might be missing after cleaning
df.dropna(subset=['price', 'service fee'], inplace=True)

print("'price' and 'service fee' columns cleaned and converted to numeric.")

# --- Final Check after Cleaning ---
print("\n--- DataFrame Info After Cleaning ---")
df.info()

print("\n--- Missing Values After Cleaning ---")
print(df.isnull().sum())


# 3.1 Plot a histogram of the 'price' column.
# We'll filter out extreme outliers for a more informative plot.
plt.figure(figsize=(12, 6))
sns.histplot(df[df['price'] < 1200]['price'], bins=50, kde=True)
plt.title('Distribution of Airbnb Prices (Capped at $1200)', fontsize=16)
plt.xlabel('Price ($)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()

# 3.2 Create a bar plot showing the count of listings in each 'neighbourhood group'.
plt.figure(figsize=(12, 6))
sns.countplot(y='neighbourhood group', data=df, order=df['neighbourhood group'].value_counts().index)
plt.title('Number of Listings by Neighbourhood Group', fontsize=16)
plt.xlabel('Count of Listings', fontsize=12)
plt.ylabel('Neighbourhood Group', fontsize=12)
plt.show()

# 3.3 Create a scatter plot of 'price' vs. 'number of reviews'.
plt.figure(figsize=(12, 6))
sns.scatterplot(x='number of reviews', y='price', data=df[df['price'] < 1200], alpha=0.5)
plt.title('Price vs. Number of Reviews', fontsize=16)
plt.xlabel('Number of Reviews', fontsize=12)
plt.ylabel('Price ($)', fontsize=12)
plt.show()

# 3.4 Create a box plot of 'price' by 'room type'.
plt.figure(figsize=(12, 6))
sns.boxplot(x='room type', y='price', data=df[df['price'] < 1200])
plt.title('Price Distribution by Room Type', fontsize=16)
plt.xlabel('Room Type', fontsize=12)
plt.ylabel('Price ($)', fontsize=12)
plt.show()

# 3.5 Calculate and display the correlation matrix for the numerical columns.
numeric_cols = df.select_dtypes(include=np.number)
correlation_matrix = numeric_cols.corr()
print("\n--- Correlation Matrix ---")
display(correlation_matrix)

# 3.6 Plot a heatmap of the correlation matrix.
plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Heatmap of Correlation Matrix', fontsize=16)
plt.show()


# 4.1 What is the average price of listings in each neighbourhood group?
avg_price_by_neighbourhood = df.groupby('neighbourhood group')['price'].mean().sort_values(ascending=False)
print("--- Average Price by Neighbourhood Group ---")
print(avg_price_by_neighbourhood)
print("-" * 40)

# 4.2 Which host has the most listings?
top_host = df['host name'].value_counts().nlargest(1)
print("\n--- Host with the Most Listings ---")
print(top_host)
print("-" * 40)

# 4.3 What are the top 5 most reviewed listings?
top_5_reviewed = df.nlargest(5, 'number of reviews')[['NAME', 'number of reviews', 'host name', 'price']]
print("\n--- Top 5 Most Reviewed Listings ---")
display(top_5_reviewed)
print("-" * 40)

# 4.4 How does the price vary by room type?
price_by_room_type_summary = df.groupby('room type')['price'].describe()
print("\n--- Price Variation by Room Type (Summary Statistics) ---")
display(price_by_room_type_summary)
print("The box plot in the EDA section also provides a great visual answer to this question.")
print("-" * 40)

# 4.5 What is the distribution of cancellation policies?
cancellation_policy_dist = df['cancellation_policy'].value_counts()
print("\n--- Distribution of Cancellation Policies ---")
print(cancellation_policy_dist)

# Plotting the distribution
plt.figure(figsize=(10, 6))
sns.countplot(x='cancellation_policy', data=df, order=cancellation_policy_dist.index)
plt.title('Distribution of Cancellation Policies', fontsize=16)
plt.xlabel('Cancellation Policy', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```
